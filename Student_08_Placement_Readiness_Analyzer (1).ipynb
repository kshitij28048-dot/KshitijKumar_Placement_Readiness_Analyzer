## 2. Load your dataset

```python
df = pd.read_csv("Student_08_dataset.csv")
print("First Five Records : ")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
print("Dataset Shape (Rows,Columns) : ",df.shape)
print("Columns Name:", list(df.columns))

df.info()


# TODO: Inspect data types and missing values
print("Missing Vaules in each columns: ")
display(df.isnull().sum())


# TODO: Count duplicate rows
print("Number of Duplicate Rows : ",df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
col=["Aptitude","Reasoning","Verbal","Technical"]

# TODO: Convert them to a NumPy array using .to_numpy()
arr=df[col].to_numpy()

# TODO: Calculate average, maximum and minimum
print("Mean : ",np.mean(arr))
print("Maximum : ",np.max(arr))
print("Minimum : ",np.min(arr))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
df["Total Marks"]=df[col].sum(axis=1)
df["Percentage"]=df["Total Marks"]/len(col)
display(df[["Student_ID","Name"]+col+["Total Marks","Percentage"]])


# TODO: Sort records and display the top/bottom results
sorted_df=df.sort_values(by="Percentage",ascending=False)
topper=df.sort_values(by="Percentage",ascending=False).head(1)
display(topper[["Name","Total Marks","Percentage"]])
display(sorted_df)


# TODO: Filter records using a condition

df["Result"]=df[col].ge(50).all(axis=1).map({
    True: "Passes",
    False: "Failed"})
display(df[["Name","Total Marks","Percentage","Result"]])

    

    
    
    

```
